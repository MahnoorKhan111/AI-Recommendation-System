## Project Objective

The objective of this project is to build a simple recommendation
system that:

1. Takes user preferences as input.
2. Compares user preferences with item attributes.
3. Calculates a similarity score.
4. Ranks items according to similarity.
5. Displays the most relevant recommendations.

```python
movies = [
    {
        "title": "Interstellar",
        "genres": ["Sci-Fi", "Adventure", "Drama"]
    },
    {
        "title": "Inception",
        "genres": ["Sci-Fi", "Action", "Thriller"]
    },
    {
        "title": "Avengers",
        "genres": ["Action", "Adventure", "Sci-Fi"]
    },
    {
        "title": "The Dark Knight",
        "genres": ["Action", "Crime", "Drama"]
    },
    {
        "title": "Toy Story",
        "genres": ["Animation", "Comedy", "Family"]
    },
    {
        "title": "Finding Nemo",
        "genres": ["Animation", "Adventure", "Family"]
    },
    {
        "title": "The Matrix",
        "genres": ["Sci-Fi", "Action", "Thriller"]
    }
]

print("Number of movies:", len(movies))
```

```python
user_preferences = ["Sci-Fi", "Adventure"]

print("User Preferences:", user_preferences)
```

```python
def calculate_similarity(user_preferences, movie_genres):
    matched_genres = set(user_preferences).intersection(set(movie_genres))

    similarity_score = len(matched_genres) / len(user_preferences)

    return similarity_score
```

```python
score = calculate_similarity(
    ["Sci-Fi", "Adventure"],
    ["Sci-Fi", "Adventure", "Drama"]
)

print("Similarity Score:", score)
print("Similarity Percentage:", score * 100, "%")
```

```python
recommendations = []

for movie in movies:
    score = calculate_similarity(
        user_preferences,
        movie["genres"]
    )

    recommendations.append({
        "title": movie["title"],
        "score": score
    })

print(recommendations)
```

```python
recommendations = sorted(
    recommendations,
    key=lambda x: x["score"],
    reverse=True
)

print(recommendations)
```

```python
print("======================================")
print("       AI MOVIE RECOMMENDATION")
print("======================================")

print("\nYour Interests:")
for preference in user_preferences:
    print("-", preference)

print("\nRecommended Movies:")

for i, movie in enumerate(recommendations[:5], start=1):
    percentage = movie["score"] * 100

    print(
        f"{i}. {movie['title']} - "
        f"Match Score: {percentage:.1f}%"
    )
```

```python
# ============================================
# AI MOVIE RECOMMENDATION SYSTEM
# Project 3 - DecodeLabs
# ============================================

# Movie dataset
movies = [
    {
        "title": "Interstellar",
        "genres": ["Sci-Fi", "Adventure", "Drama"]
    },
    {
        "title": "Inception",
        "genres": ["Sci-Fi", "Action", "Thriller"]
    },
    {
        "title": "Avengers",
        "genres": ["Action", "Adventure", "Sci-Fi"]
    },
    {
        "title": "The Dark Knight",
        "genres": ["Action", "Crime", "Drama"]
    },
    {
        "title": "Toy Story",
        "genres": ["Animation", "Comedy", "Family"]
    },
    {
        "title": "Finding Nemo",
        "genres": ["Animation", "Adventure", "Family"]
    },
    {
        "title": "The Matrix",
        "genres": ["Sci-Fi", "Action", "Thriller"]
    }
]

# Similarity function
def calculate_similarity(user_preferences, movie_genres):

    matched_genres = set(user_preferences).intersection(
        set(movie_genres)
    )

    similarity_score = (
        len(matched_genres) / len(user_preferences)
    )

    return similarity_score


# Display available genres
print("======================================")
print("       AI MOVIE RECOMMENDATION")
print("======================================")

print("\nAvailable Genres:")

genres = [
    "Action",
    "Adventure",
    "Sci-Fi",
    "Drama",
    "Comedy",
    "Animation",
    "Family",
    "Thriller",
    "Crime"
]

for genre in genres:
    print("-", genre)


# Get user preferences
user_input = input(
    "\nEnter your favorite genres separated by commas: "
)

user_preferences = [
    genre.strip().title()
    for genre in user_input.split(",")
]


# Calculate recommendations
recommendations = []

for movie in movies:

    score = calculate_similarity(
        user_preferences,
        movie["genres"]
    )

    recommendations.append({
        "title": movie["title"],
        "score": score
    })


# Sort by similarity score
recommendations = sorted(
    recommendations,
    key=lambda x: x["score"],
    reverse=True
)


# Display recommendations
print("\n======================================")
print("          RECOMMENDATIONS")
print("======================================")

print("\nYour Preferences:")

for preference in user_preferences:
    print("-", preference)


print("\nTop Recommended Movies:")

for i, movie in enumerate(
    recommendations[:5],
    start=1
):

    percentage = movie["score"] * 100

    print(
        f"{i}. {movie['title']} "
        f"- Match Score: {percentage:.1f}%"
    )
```
